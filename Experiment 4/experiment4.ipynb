## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

## 2. Load and prepare the Superstore dataset

```python
url = "https://raw.githubusercontent.com/leonism/sample-superstore/master/data/superstore.csv"
df = pd.read_csv(url, encoding="latin1")
print("Shape:", df.shape)
display(df.head())
for col in ["Order Date", "Ship Date"]:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")
print("Missing values:")
display(df.isnull().sum().sort_values(ascending=False).head(15))
display(df.describe(include="all").T)
```

## 3. Univariate visualizations

```python
for col in ["Sales", "Profit", "Quantity", "Discount"]:
    if col in df.columns:
        plt.figure(figsize=(8, 4))
        sns.histplot(df[col].dropna(), kde=True, bins=30)
        plt.title(f"Distribution of {col}")
        plt.tight_layout()
        plt.show()
```

## 4. Category and regional analysis

```python
if {"Category", "Sales"}.issubset(df.columns):
    plt.figure(figsize=(8, 5))
    sns.barplot(data=df, x="Category", y="Sales", estimator="sum", errorbar=None)
    plt.title("Total Sales by Category")
    plt.tight_layout()
    plt.show()
if {"Region", "Profit"}.issubset(df.columns):
    region_profit = df.groupby("Region", as_index=False)["Profit"].sum().sort_values("Profit", ascending=False)
    sns.barplot(data=region_profit, x="Region", y="Profit")
    plt.title("Total Profit by Region")
    plt.show()
if {"Sub-Category", "Sales"}.issubset(df.columns):
    sub_sales = df.groupby("Sub-Category", as_index=False)["Sales"].sum().sort_values("Sales", ascending=False)
    sns.barplot(data=sub_sales, y="Sub-Category", x="Sales")
    plt.title("Sales by Sub-Category")
    plt.tight_layout()
    plt.show()
```

## 5. Time-series visualization

```python
if {"Order Date", "Sales"}.issubset(df.columns):
    monthly_sales = (df.dropna(subset=["Order Date"])
                     .set_index("Order Date")["Sales"].resample("MS").sum())
    monthly_sales.plot(figsize=(12, 5), marker="o")
    plt.title("Monthly Sales Trend")
    plt.xlabel("Order Month")
    plt.ylabel("Sales")
    plt.tight_layout()
    plt.show()
```

## 6. Correlation and outlier analysis

```python
numeric = df.select_dtypes(include=np.number)
plt.figure(figsize=(10, 7))
sns.heatmap(numeric.corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Numeric Feature Correlation")
plt.tight_layout()
plt.show()
for col in ["Sales", "Profit", "Discount"]:
    if col in df.columns:
        plt.figure(figsize=(8, 4))
        sns.boxplot(x=df[col])
        plt.title(f"Outliers in {col}")
        plt.show()
```
