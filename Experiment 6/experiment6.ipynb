## Aim
Train a Linear Regression model on the Scikit-learn diabetes dataset and evaluate regression performance.

> Run cells from top to bottom. Dataset sources and any synthetic-data notes are documented in the relevant sections.

## 1. Import libraries and load dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

data = load_diabetes()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name="target")
df = X.copy()
df["target"] = y
print("Dataset shape:", df.shape)
display(df.head())
```

## 2. Inspect data

```python
print("Columns:", df.columns.tolist())
print("Data types:")
print(df.dtypes)
print("Missing values:")
display(df.isnull().sum())
display(df.describe())
```

## 3. Feature/target selection and train-test split

```python
X = df.drop(columns=["target"])
y = df["target"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 4. Train and predict

```python
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Model trained. Intercept:", model.intercept_)
comparison = pd.DataFrame({"Actual": y_test.values, "Predicted": y_pred})
display(comparison.head(10))
```

## 5. Actual vs predicted and regression line

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.75)
minimum, maximum = min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())
plt.plot([minimum, maximum], [minimum, maximum], linewidth=2)
plt.title("Actual vs Predicted Values")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.tight_layout()
plt.show()

feature = "bmi"
simple_model = LinearRegression().fit(X_train[[feature]], y_train)
feature_values = np.linspace(X[feature].min(), X[feature].max(), 100).reshape(-1, 1)
plt.figure(figsize=(9, 5))
plt.scatter(X[feature], y, alpha=0.5)
plt.plot(feature_values, simple_model.predict(feature_values), linewidth=2)
plt.title("Regression Line: BMI vs Target")
plt.xlabel("BMI")
plt.ylabel("Target")
plt.tight_layout()
plt.show()
```

## 6. Evaluate model performance

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
display(pd.DataFrame({"Metric": ["MAE", "MSE", "RMSE", "R²"], "Value": [mae, mse, rmse, r2]}))
print(f"MAE={mae:.4f}, MSE={mse:.4f}, RMSE={rmse:.4f}, R²={r2:.4f}")
```

## 7. Coefficients and residual analysis

```python
coefficients = pd.DataFrame({"Feature": X.columns, "Coefficient": model.coef_})
coefficients["Absolute Impact"] = coefficients["Coefficient"].abs()
coefficients = coefficients.sort_values("Absolute Impact", ascending=False)
display(coefficients)
coefficients.sort_values("Coefficient").plot(kind="barh", x="Feature", y="Coefficient", legend=False, figsize=(9, 6))
plt.title("Linear Regression Coefficients")
plt.tight_layout()
plt.show()

residuals = y_test.values - y_pred
plt.figure(figsize=(9, 5))
plt.scatter(y_pred, residuals, alpha=0.75)
plt.axhline(0, linewidth=2)
plt.title("Residual Plot")
plt.xlabel("Predicted Values")
plt.ylabel("Residuals")
plt.tight_layout()
plt.show()
print("Residual mean:", residuals.mean())
```
