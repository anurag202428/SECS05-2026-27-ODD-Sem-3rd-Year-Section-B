## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

## 2. Load IBM HR dataset

```python
url = "https://raw.githubusercontent.com/shantanu1109/IBM-HR-Analytics-Employee-Attrition-and-Performance-Prediction/main/DATASET/WA_Fn-UseC_-HR-Employee-Attrition.csv"
df = pd.read_csv(url)
print("Shape:", df.shape)
display(df.head())
```

## 3. Explore structure and descriptive statistics

```python
print("Columns:", df.columns.tolist())
df.info()
display(df.describe(include="all").T)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

## 4. Numerical distributions

```python
selected_numeric = [c for c in ["Age", "MonthlyIncome", "JobLevel", "YearsAtCompany"] if c in df.columns]
df[selected_numeric].hist(figsize=(12, 8), bins=20)
plt.suptitle("Distribution of Numerical Features")
plt.tight_layout()
plt.show()

if "MonthlyIncome" in df.columns:
    plt.figure(figsize=(8, 5))
    sns.kdeplot(data=df, x="MonthlyIncome", fill=True)
    plt.title("Distribution of Monthly Income")
    plt.show()
```

## 5. Categorical analysis

```python
categorical_columns = [c for c in ["Department", "JobRole", "Gender", "Attrition"] if c in df.columns]
for col in categorical_columns:
    plt.figure(figsize=(9, 5))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.title(f"Count Plot of {col}")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.show()
    display(df[col].value_counts())
```

## 6. Correlation, outliers, and relationships

```python
numeric_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(14, 10))
sns.heatmap(numeric_df.corr(), cmap="coolwarm", center=0, linewidths=0.3)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

box_cols = [c for c in ["Age", "MonthlyIncome", "TotalWorkingYears", "YearsAtCompany"] if c in df.columns]
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[box_cols])
plt.title("Selected Numerical Features")
plt.show()

if {"MonthlyIncome", "TotalWorkingYears"}.issubset(df.columns):
    sns.scatterplot(data=df, x="TotalWorkingYears", y="MonthlyIncome", hue="Attrition")
    plt.title("Monthly Income vs Total Working Years")
    plt.show()
```

## 7. Univariate and bivariate analysis

```python
for col in ["Age", "MonthlyIncome"]:
    if col in df.columns:
        print(col, "mean:", round(df[col].mean(), 2), "median:", round(df[col].median(), 2))
if "Attrition" in df.columns:
    display(df["Attrition"].value_counts())
if {"Attrition", "Department"}.issubset(df.columns):
    attrition_department = pd.crosstab(df["Department"], df["Attrition"], normalize="index") * 100
    display(attrition_department.round(2))
    attrition_department.plot(kind="bar", figsize=(9, 5))
    plt.title("Attrition Percentage by Department")
    plt.ylabel("Percentage")
    plt.tight_layout()
    plt.show()
```
