## 1. Imports and dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
sns.set_theme(style="whitegrid")

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
data = df[["survived", "age", "fare", "pclass"]].dropna()
display(data.head())
```

## 2. Descriptive statistics

```python
for label, result in [
    ("Mean", data[["age", "fare"]].mean()),
    ("Median", data[["age", "fare"]].median()),
    ("Mode", data[["age", "fare"]].mode().iloc[0]),
    ("Variance", data[["age", "fare"]].var()),
    ("Standard deviation", data[["age", "fare"]].std()),
]:
    print(label)
    display(result)
display(data[["age", "fare"]].describe())
```

## 3. Pearson correlation

```python
r, p_corr = stats.pearsonr(data["age"], data["fare"])
print("Pearson correlation:", round(r, 4))
print("p-value:", round(p_corr, 6))
correlation_matrix = data[["age", "fare", "pclass"]].corr()
display(correlation_matrix)
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", center=0)
plt.title("Pearson Correlation Matrix")
plt.show()
```

## 4. Hypothesis test: fare by survival

```python
survived_fare = data.loc[data["survived"] == 1, "fare"]
not_survived_fare = data.loc[data["survived"] == 0, "fare"]
print("Mean fare of survivors:", round(survived_fare.mean(), 2))
print("Mean fare of non-survivors:", round(not_survived_fare.mean(), 2))

t_stat, p_value = stats.ttest_ind(survived_fare, not_survived_fare, equal_var=False)
alpha = 0.05
print("t-statistic:", round(t_stat, 4), "p-value:", round(p_value, 6))
print("Reject H0" if p_value < alpha else "Fail to reject H0")
```

## 5. One-way ANOVA and group visualization

```python
class_fares = [data.loc[data["pclass"] == c, "fare"] for c in [1, 2, 3]]
f_stat, p_anova = stats.f_oneway(*class_fares)
print("ANOVA F-statistic:", round(f_stat, 4), "p-value:", round(p_anova, 6))
sns.boxplot(data=data, x="pclass", y="fare")
plt.title("Fare Distribution Across Passenger Classes")
plt.show()
```

## 6. Simple linear regression

```python
X = sm.add_constant(data["age"])
y = data["fare"]
model = sm.OLS(y, X).fit()
print(model.summary())
print("Intercept:", round(model.params["const"], 4))
print("Age coefficient:", round(model.params["age"], 4))
print("R²:", round(model.rsquared, 4))
print("Adjusted R²:", round(model.rsquared_adj, 4))
print("95% confidence intervals:")
display(model.conf_int())
```
